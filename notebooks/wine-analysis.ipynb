```python

# ============================================================
# IMPORT LIBRARIES
# ============================================================

# MLflow
import mlflow
import mlflow.sklearn

# Data manipulation
import pandas as pd
import numpy as np

# Visualization
import matplotlib
matplotlib.use("Agg")  # Prevent Tkinter errors

import matplotlib.pyplot as plt
import seaborn as sns

# Machine Learning
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score
from sklearn.metrics import classification_report
from sklearn.metrics import confusion_matrix

# Save model
import joblib
```

```python
# ============================================================
# LOAD DATASET
# ============================================================

wine_dataset = pd.read_csv("../data/raw/wine.csv")

wine_dataset.head()
```

```python
# ============================================================
# DATASET SHAPE
# ============================================================

print("Dataset Shape")
print("------------------------")

print(wine_dataset.shape)
```

```python
# ============================================================
# FIRST FIVE ROWS
# ============================================================

wine_dataset.head()
```

```python
# ============================================================
# DATASET INFORMATION
# ============================================================

wine_dataset.info()
```

```python
# ============================================================
# CHECK MISSING VALUES
# ============================================================

wine_dataset.isnull().sum()
     
```

```python
# ============================================================
# DESCRIPTIVE STATISTICS
# ============================================================

wine_dataset.describe()
```

```python
# ============================================================
# WINE QUALITY DISTRIBUTION
# ============================================================

plt.figure(figsize=(8,5))

sns.countplot(
    x="quality",
    data=wine_dataset
)

plt.title("Wine Quality Distribution")

plt.show()
     
```

```python
# ============================================================
# VOLATILE ACIDITY VS QUALITY
# ============================================================

plt.figure(figsize=(8,5))

sns.barplot(
    x="quality",
    y="volatile acidity",
    data=wine_dataset
)

plt.title("Volatile Acidity vs Quality")

plt.show()
```

```python
# ============================================================
# CORRELATION HEATMAP
# ============================================================

correlation = wine_dataset.corr()

plt.figure(figsize=(12,10))

sns.heatmap(
    correlation,
    annot=True,
    fmt=".2f",
    cmap="coolwarm"
)

plt.title("Wine Correlation Heatmap")

plt.show()
```

```python
# ============================================================
# FEATURE ENGINEERING
# ============================================================

X = wine_dataset.drop("quality", axis=1)

Y = wine_dataset["quality"].apply(
    lambda y: 1 if y >= 7 else 0
)

print(X.head())

print(Y.head())
```

```python
# ========================================================
# TRAIN TEST SPLIT
# ========================================================

test_size = 0.2
random_state = 2

X_train, X_test, Y_train, Y_test = train_test_split(
        X,
        Y,
        test_size=test_size,
        random_state=random_state
    )
```

```python
# ========================================================
# MODEL CREATION
# ========================================================

n_estimators = 100

model = RandomForestClassifier(
        n_estimators=n_estimators,
        random_state=42
    )
```

```python
# ========================================================
# MODEL TRAINING
# ========================================================

model.fit(
        X_train,
        Y_train
    )
```

```python
# ========================================================
    # PREDICTIONS
    # ========================================================

predictions = model.predict(
        X_test
    )
```

```python
# ========================================================
# EVALUATION
# ========================================================

accuracy = accuracy_score(
        Y_test,
        predictions
    )

print("\nModel Accuracy:")
print(accuracy)
```

```python
# ========================================================
# CLASSIFICATION REPORT
# ========================================================

report = classification_report(
        Y_test,
        predictions
    )

print("\nClassification Report:")
print(report)

with open(
        "classification_report.txt",
        "w"
    ) as file:
        file.write(report)
```

```python
# ========================================================
    # CONFUSION MATRIX
    # ========================================================

cm = confusion_matrix(
        Y_test,
        predictions
    )

plt.figure(figsize=(6, 5))

sns.heatmap(
        cm,
        annot=True,
        fmt="d"
    )

plt.title("Confusion Matrix")

plt.savefig(
        "confusion_matrix.png",
        bbox_inches="tight"
    )

mlflow.log_artifact(
        "confusion_matrix.png",
        artifact_path="plots"
    )

plt.close()
```

```python

    # ========================================================
    # SAVE MODEL USING JOBLIB
    # ========================================================

joblib.dump(
        model,
        "wine_quality.pkl"
    )

mlflow.log_artifact(
        "wine_quality.pkl",
        artifact_path="saved_models"
    )
```

```python
# ========================================================
# SAMPLE PREDICTION
# ========================================================

input_data = (
        7.3,
        0.65,
        0.00,
        1.2,
        0.065,
        15.0,
        21.0,
        0.9966,
        3.2,
        0.68,
        10.0
    )

input_array = np.asarray(
        input_data
    ).reshape(1, -1)

prediction = model.predict(
        input_array
    )

print("\nPrediction Result:")

if prediction[0] == 1:
        print("Good Quality Wine")
else:
        print("Bad Quality Wine")

print("\nMLflow Run Completed Successfully")
```
